```python
#making an copy of the dataset
df = data.copy()
df.shape
```

```python
#converting numeric target (yes = 1, no = 0)
df['is_client_subscribed'] = df['y'].apply(lambda x: 1 if x == 'yes' else 0)
```

```python
#channel mapping (contact column)
df['Marketing_Channel'] = df['contact'].map({
    'cellular': 'Mobile Outreach',
    'telephone': 'Landline Outreach',
    'unknown': 'Unclassified Channel'
})
```

```python
#funnel stage identification
df['Engaged_Lead'] = df['duration'].apply(lambda x: 1 if x > 60 else 0)
df['Qualified_Lead'] = np.where((df['balance'] > 0) & (df['default'] == 'no'), 1, 0)
```

```python
#age group segments
def age_bracket(age):
    if age < 30: return "18-29 (Young Adult)"
    elif age <= 50: return "30-50 (Working Class)"
    else: return "50+ (Senior/Retirement)"

df['Age_Group'] = df['age'].apply(age_bracket)
```

```python
#call duration category
df['Call_Duration_Min'] = df['duration'] / 60
```

```python
df.to_csv("Updated_bank_data.csv",index=False)
```
